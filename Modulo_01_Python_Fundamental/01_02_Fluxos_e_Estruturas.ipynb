## 1. `while` — repita enquanto

```python
while condicao:
    # corpo, executado enquanto a condição for verdadeira
```

O `while` testa **antes** de cada iteração. Se a condição já começa falsa, o corpo nunca roda.

**Regra de ouro:** todo `while` precisa de algo no corpo que eventualmente torne a condição falsa. Senão você criou um laço infinito. (No VS Code, o botão ⏹ **Interrupt** para o kernel travado.)

````python
# Contagem regressiva de estoque
estoque = 5

while estoque > 0:
    print(f"Vendido! Restam {estoque - 1} unidades.")
    estoque -= 1

print("Produto esgotado.")
````

````python
# Acumulador: somando vendas até bater a meta
meta = 10_000
acumulado = 0
dia = 0
venda_diaria = 1_850

while acumulado < meta:
    dia += 1
    acumulado += venda_diaria
    print(f"Dia {dia:>2}: acumulado R$ {acumulado:>9,.2f}")

print(f"\nMeta de R$ {meta:,.2f} batida no dia {dia}.")
````

### `while True` + `break` — o laço de menu

Quando você não sabe de antemão quantas iterações precisa, o padrão idiomático é um laço infinito com uma saída explícita.

Na célula abaixo simulamos as entradas com uma lista, para não travar o notebook pedindo `input()`.

````python
# Simulando as escolhas do usuário (em um script real, viria de input())
entradas_simuladas = ["1", "2", "9", "0"]
i = 0

while True:
    print("\n--- ATLAS · Menu ---")
    print("1) Relatório por cidade")
    print("2) Relatório por produto")
    print("0) Sair")

    opcao = entradas_simuladas[i]
    i += 1
    print(f">>> {opcao}")

    if opcao == "0":
        print("Encerrando. Até logo!")
        break
    elif opcao == "1":
        print("[relatório por cidade gerado]")
    elif opcao == "2":
        print("[relatório por produto gerado]")
    else:
        print("Opção inválida. Tente novamente.")
````

## 2. `for` — percorra uma coleção

O `for` do Python **não** é o `for(i=0; i<n; i++)` de C/Java. Ele é um *for-each*: percorre os elementos de qualquer **iterável** (lista, string, tupla, dicionário, arquivo...).

```python
for elemento in colecao:
    # faz algo com elemento
```

````python
cidades = ["Campinas", "São Paulo", "Ribeirão Preto", "Sorocaba"]

for cidade in cidades:
    print(f"Processando região: {cidade}")

print()

# Strings também são iteráveis
for letra in "Atlas":
    print(letra, end=" ")
````

### `range()` — sequências de números

| Chamada | Produz |
|---------|--------|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(2, 6)` | 2, 3, 4, 5 |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 |
| `range(10, 0, -1)` | 10, 9, 8, ..., 1 |

O `fim` é sempre **exclusivo**. `range` é *lazy*: não cria a lista na memória, gera os números sob demanda (por isso `print(range(5))` mostra `range(0, 5)` e não os números).

````python
print("range(5):        ", list(range(5)))
print("range(2, 6):     ", list(range(2, 6)))
print("range(0, 10, 2): ", list(range(0, 10, 2)))
print("range(10, 0, -1):", list(range(10, 0, -1)))

print()
for i in range(1, 6):
    print(f"{i} x 7 = {i * 7}")
````

### `enumerate()` e `zip()` — os dois utilitários que mudam sua vida

- **`enumerate(colecao, start=0)`** → devolve pares `(indice, elemento)`. Use quando precisa do índice **e** do valor.
- **`zip(a, b)`** → junta duas (ou mais) coleções em pares. Para quando a **mais curta** acabar.

Evite o antipadrão `for i in range(len(lista)):` — quase sempre `enumerate` é mais legível.

````python
produtos = ["Notebook", "Mouse", "Teclado", "Monitor"]
precos = [2599.90, 89.90, 249.00, 1199.00]
estoques = [12, 340, 88, 5]

# enumerate
for posicao, produto in enumerate(produtos, start=1):
    print(f"{posicao}º - {produto}")

print()

# zip
for produto, preco, estoque in zip(produtos, precos, estoques):
    valor_parado = preco * estoque
    print(f"{produto:<10} R$ {preco:>9,.2f} x {estoque:>4} un = R$ {valor_parado:>12,.2f}")
````

## 3. `break`, `continue` e o `else` do laço

- **`break`** — abandona o laço imediatamente.
- **`continue`** — pula para a próxima iteração, ignorando o resto do corpo.
- **`else` do laço** — executa **apenas se o laço terminou sem `break`**. Perfeito para buscas do tipo "procurei e não achei".

````python
pedidos = [1201, 1202, 1203, 1204, 1205]
procurado = 1203

for pedido in pedidos:
    print(f"Verificando {pedido}...")
    if pedido == procurado:
        print(f"✅ Pedido {procurado} encontrado!")
        break
else:
    print(f"❌ Pedido {procurado} não existe na base.")
````

````python
# continue: ignorando registros inválidos
valores_brutos = ["120.50", "", "89.90", "n/d", "45.00", None, "300.00"]
total = 0.0
ignorados = 0

for valor in valores_brutos:
    if not valor:                 # pega "" e None (falsy)
        ignorados += 1
        continue
    if not valor.replace(".", "").isdigit():
        ignorados += 1
        continue
    total += float(valor)

print(f"Total válido: R$ {total:,.2f}")
print(f"Registros ignorados: {ignorados}")
````

## 4. Listas

Uma `list` é uma sequência **ordenada** e **mutável**, que aceita **tipos misturados**.

```python
lista = [elemento1, elemento2, elemento3]
```

Ordenada = a posição importa e é estável. Mutável = você pode alterar, adicionar e remover elementos **no mesmo objeto**.

````python
vazia = []
cidades = ["Campinas", "São Paulo", "Sorocaba"]
mista = [1042, "Campinas", 2599.90, True, None]
aninhada = [["Campinas", 12000], ["São Paulo", 34000]]

print(cidades, len(cidades))
print(mista)
print(aninhada[1][0], "->", aninhada[1][1])
````

### Indexação e fatiamento — igual às strings

A diferença crucial: em lista o slice também funciona do **lado esquerdo** da atribuição, permitindo substituir trechos inteiros.

````python
produtos = ["Notebook", "Mouse", "Teclado", "Monitor", "Webcam"]

print("produtos[0]    =", produtos[0])
print("produtos[-1]   =", produtos[-1])
print("produtos[1:4]  =", produtos[1:4])
print("produtos[:2]   =", produtos[:2])
print("produtos[::2]  =", produtos[::2])
print("produtos[::-1] =", produtos[::-1])
````

````python
# Mutabilidade
produtos = ["Notebook", "Mouse", "Teclado"]
produtos[1] = "Mouse Gamer"
print(produtos)

produtos[0:2] = ["Ultrabook", "Mouse sem fio"]   # substitui um trecho
print(produtos)
````

### Métodos de lista

| Método | O que faz | Altera a lista? |
|--------|-----------|-----------------|
| `append(x)` | Adiciona `x` ao final | ✅ |
| `insert(i, x)` | Insere `x` na posição `i` | ✅ |
| `extend(outra)` | Concatena outra coleção | ✅ |
| `remove(x)` | Remove a **primeira** ocorrência de `x` | ✅ |
| `pop(i)` | Remove e **devolve** o item `i` (padrão: último) | ✅ |
| `clear()` | Esvazia | ✅ |
| `sort()` | Ordena **no lugar** | ✅ |
| `reverse()` | Inverte **no lugar** | ✅ |
| `index(x)` | Posição da primeira ocorrência | ❌ |
| `count(x)` | Quantas vezes `x` aparece | ❌ |
| `copy()` | Cópia **rasa** | ❌ |

⚠️ Os métodos que alteram no lugar devolvem `None`. `lista = lista.sort()` é um bug clássico — destrói sua lista.

````python
carrinho = ["Mouse"]

carrinho.append("Teclado")
carrinho.insert(0, "Notebook")
carrinho.extend(["Monitor", "Webcam"])
print("Depois de adicionar:", carrinho)

carrinho.remove("Webcam")
removido = carrinho.pop()
print(f"Removido com pop(): {removido}")
print("Estado final:", carrinho)
print("Posição de 'Teclado':", carrinho.index("Teclado"))
````

````python
# A ARMADILHA do sort() devolvendo None
numeros = [5, 2, 8, 1]

errado = numeros.sort()      # <- sort() ordena no lugar e devolve None
print("errado =", errado)
print("numeros (foi alterado) =", numeros)

# A forma correta quando você quer uma NOVA lista:
originais = [5, 2, 8, 1]
ordenados = sorted(originais)
print("\noriginais  =", originais)
print("ordenados  =", ordenados)
````

### Ordenação com `key` e `reverse`

`sorted()` e `.sort()` aceitam:

- **`key=`** — uma função aplicada a cada elemento para decidir o critério.
- **`reverse=True`** — ordem decrescente.

Isso é o que permite ordenar registros por qualquer campo.

````python
vendas = [
    ("Campinas", 45300.00),
    ("São Paulo", 128900.50),
    ("Sorocaba", 22100.00),
    ("Ribeirão Preto", 67450.75),
]

# Por nome da cidade (elemento [0])
print("--- Alfabético ---")
for cidade, valor in sorted(vendas):
    print(f"{cidade:<18} R$ {valor:>12,.2f}")

# Por faturamento decrescente (elemento [1])
print("\n--- Ranking de faturamento ---")
ranking = sorted(vendas, key=lambda item: item[1], reverse=True)
for pos, (cidade, valor) in enumerate(ranking, start=1):
    print(f"{pos}º {cidade:<18} R$ {valor:>12,.2f}")
````

````python
# key com funções prontas
nomes = ["ana", "Carlos", "beatriz", "Daniel"]

print(sorted(nomes))                  # maiúsculas vêm antes (ordem ASCII)
print(sorted(nomes, key=str.lower))   # ordem alfabética "humana"
print(sorted(nomes, key=len))         # por comprimento
````

### Funções embutidas que operam sobre listas

````python
valores = [45300.00, 128900.50, 22100.00, 67450.75, 91200.00]

print("Quantidade :", len(valores))
print("Soma       :", f"R$ {sum(valores):,.2f}")
print("Máximo     :", f"R$ {max(valores):,.2f}")
print("Mínimo     :", f"R$ {min(valores):,.2f}")
print("Média      :", f"R$ {sum(valores) / len(valores):,.2f}")

print()
print("Algum acima de 100k?", any(v > 100_000 for v in valores))
print("Todos acima de 10k? ", all(v > 10_000 for v in valores))
````

### ⚠️ Armadilha: cópia rasa vs referência

`nova = antiga` **não copia** — cria outro nome para a mesma lista. Alterar uma altera a outra.

Para copiar de verdade: `antiga.copy()`, `antiga[:]` ou `list(antiga)`.

Atenção: essas são cópias **rasas**. Se a lista contém outras listas, os elementos internos continuam compartilhados. Para independência total use `copy.deepcopy()`.

````python
original = ["Notebook", "Mouse"]

referencia = original          # MESMO objeto
copia = original.copy()        # objeto NOVO

referencia.append("Teclado")

print("original   :", original)     # foi alterado!
print("referencia :", referencia)
print("copia      :", copia)        # intacta
print()
print("original is referencia?", original is referencia)
print("original is copia?     ", original is copia)
````

````python
# Cópia rasa NÃO protege elementos aninhados
import copy

matriz = [["Campinas", 100], ["São Paulo", 200]]

rasa = matriz.copy()
profunda = copy.deepcopy(matriz)

matriz[0][1] = 999

print("matriz   :", matriz)
print("rasa     :", rasa)        # afetada!
print("profunda :", profunda)    # protegida
````

## 5. Tuplas

Uma `tuple` é uma sequência **ordenada** e **imutável**. Sintaxe: parênteses (ou nada, só vírgulas).

**Quando usar tupla em vez de lista?**

| Use **lista** quando | Use **tupla** quando |
|----------------------|----------------------|
| A coleção vai crescer/encolher | O conteúdo é fixo |
| São itens homogêneos ("os produtos") | É um **registro** ("um pedido: id, cidade, valor") |
| Você precisa ordenar no lugar | Você quer usar como chave de dicionário |

Tuplas são levemente mais rápidas e, por serem imutáveis, são **hasheáveis** — podem ser chave de `dict` ou elemento de `set`. Listas não podem.

````python
pedido = (1042, "Campinas", 2599.90, "pago")

print(pedido)
print("ID:", pedido[0], "| Cidade:", pedido[1])
print("Tamanho:", len(pedido))

# pedido[2] = 3000   # <- descomente: TypeError, tupla é imutável

# ⚠️ Tupla de 1 elemento PRECISA da vírgula:
nao_e_tupla = ("Campinas")
e_tupla = ("Campinas",)
print()
print(type(nao_e_tupla).__name__, type(e_tupla).__name__)
````

````python
# Tupla como chave de dicionário — impossível com lista
vendas_por_cidade_mes = {
    ("Campinas", "2026-07"): 45300.00,
    ("Campinas", "2026-08"): 51200.00,
    ("São Paulo", "2026-08"): 128900.50,
}

print(vendas_por_cidade_mes[("Campinas", "2026-08")])

for (cidade, mes), valor in vendas_por_cidade_mes.items():
    print(f"{cidade:<12} {mes}  R$ {valor:>12,.2f}")
````

## 6. Desempacotamento (*unpacking*)

Atribuir os elementos de uma sequência a vários nomes de uma vez. É um dos idiomas mais característicos de Python.

````python
pedido = (1042, "Campinas", 2599.90, "pago")

# Desempacotamento posicional
id_pedido, cidade, valor, status = pedido
print(id_pedido, cidade, valor, status)

# O _ é a convenção para "não me interessa este valor"
id_pedido, _, valor, _ = pedido
print(f"Pedido {id_pedido}: R$ {valor:,.2f}")
````

````python
# Desempacotamento estendido com *
valores = [120, 340, 89, 560, 45, 900, 210]

primeiro, *meio, ultimo = valores
print("primeiro:", primeiro)
print("meio    :", meio)
print("ultimo  :", ultimo)

topo, *resto = sorted(valores, reverse=True)
print(f"\nMaior venda: {topo} | demais: {resto}")
````

````python
# Desempacotamento no for — extremamente comum
registros = [
    ("Campinas", 45300.00, 312),
    ("São Paulo", 128900.50, 890),
    ("Sorocaba", 22100.00, 156),
]

print(f"{'Cidade':<14}{'Faturamento':>15}{'Pedidos':>10}{'Ticket':>12}")
print("-" * 51)
for cidade, faturamento, pedidos in registros:
    ticket = faturamento / pedidos
    print(f"{cidade:<14}{faturamento:>15,.2f}{pedidos:>10}{ticket:>12,.2f}")
````

## 7. Laços aninhados e dados tabulares

Um laço dentro de outro. É assim que se percorre uma **matriz** (lista de listas) — a representação mais crua de uma planilha.

A regra: o laço interno completa **todas** as suas voltas para **cada** volta do externo.

````python
# Tabuada — o exemplo clássico de laço aninhado
for i in range(1, 4):
    for j in range(1, 6):
        print(f"{i}x{j}={i*j:<3}", end=" ")
    print()   # quebra de linha ao fim de cada linha da tabuada
````

````python
# Matriz de vendas: linhas = cidades, colunas = meses
meses = ["Jun", "Jul", "Ago"]
dados = [
    ["Campinas",       41200.00, 45300.00, 51200.00],
    ["São Paulo",     110500.00, 118700.00, 128900.50],
    ["Sorocaba",       19800.00, 22100.00, 20450.00],
    ["Ribeirão Preto", 58300.00, 67450.75, 71100.00],
]

# Cabeçalho
print(f"{'Cidade':<16}" + "".join(f"{m:>14}" for m in meses) + f"{'TOTAL':>15}")
print("-" * 75)

total_geral = 0.0
totais_mes = [0.0, 0.0, 0.0]

for linha in dados:
    cidade = linha[0]
    valores = linha[1:]
    total_cidade = sum(valores)
    total_geral += total_cidade

    for idx, v in enumerate(valores):
        totais_mes[idx] += v

    print(f"{cidade:<16}" + "".join(f"{v:>14,.2f}" for v in valores) + f"{total_cidade:>15,.2f}")

print("-" * 75)
print(f"{'TOTAL':<16}" + "".join(f"{t:>14,.2f}" for t in totais_mes) + f"{total_geral:>15,.2f}")
````

## 🔧 Prática guiada — Relatório de vendas por cidade

Este é literalmente o problema do **Módulo 01 do projeto Atlas**: *"Ninguém sabe quanto vendemos por cidade"*.

Aqui os dados estão fixos no código. No projeto, virão de um CSV.

````python
# Cada tupla: (id_pedido, cidade, produto, quantidade, preco_unitario, status)
pedidos = [
    (1001, "Campinas",       "Notebook", 2, 2599.90, "pago"),
    (1002, "São Paulo",      "Mouse",   10,   89.90, "pago"),
    (1003, "Campinas",       "Teclado",  3,  249.00, "cancelado"),
    (1004, "Sorocaba",       "Monitor",  1, 1199.00, "pago"),
    (1005, "São Paulo",      "Notebook", 1, 2599.90, "pago"),
    (1006, "Campinas",       "Monitor",  4, 1199.00, "pago"),
    (1007, "Ribeirão Preto", "Mouse",   25,   89.90, "pago"),
    (1008, "São Paulo",      "Teclado",  6,  249.00, "pendente"),
    (1009, "Campinas",       "Notebook", 1, 2599.90, "pago"),
    (1010, "Sorocaba",       "Mouse",    8,   89.90, "pago"),
]

# --- 1. Filtrar apenas pedidos faturados ---
faturados = []
for p in pedidos:
    if p[5] == "pago":
        faturados.append(p)

print(f"Pedidos recebidos: {len(pedidos)} | faturados: {len(faturados)}")

# --- 2. Agrupar por cidade (sem dicionário ainda — só listas!) ---
cidades = []
for _, cidade, _, _, _, _ in faturados:
    if cidade not in cidades:
        cidades.append(cidade)

resultado = []
for cidade in cidades:
    total = 0.0
    qtd_pedidos = 0
    itens = 0
    for _, c, _, quantidade, preco, _ in faturados:
        if c == cidade:
            total += quantidade * preco
            qtd_pedidos += 1
            itens += quantidade
    resultado.append((cidade, total, qtd_pedidos, itens))

# --- 3. Ordenar por faturamento decrescente ---
resultado.sort(key=lambda r: r[1], reverse=True)

# --- 4. Imprimir o relatório ---
faturamento_total = sum(r[1] for r in resultado)

print()
print("=" * 72)
print(f"{'ATLAS · FATURAMENTO POR CIDADE':^72}")
print("=" * 72)
print(f"{'#':<3}{'Cidade':<18}{'Faturamento':>16}{'Pedidos':>10}{'Itens':>8}{'Share':>9}")
print("-" * 72)

for pos, (cidade, total, qtd, itens) in enumerate(resultado, start=1):
    share = total / faturamento_total
    print(f"{pos:<3}{cidade:<18}{total:>16,.2f}{qtd:>10}{itens:>8}{share:>9.1%}")

print("-" * 72)
print(f"{'':<3}{'TOTAL':<18}{faturamento_total:>16,.2f}")
print("=" * 72)
print(f"\n🏆 Melhor praça: {resultado[0][0]} (R$ {resultado[0][1]:,.2f})")
print(f"📉 Menor praça:  {resultado[-1][0]} (R$ {resultado[-1][1]:,.2f})")
````
