## 1. Variáveis: apelidos para objetos

Em muitas linguagens uma variável é uma "caixinha" onde você guarda um valor. **Em Python isso é falso e entender a diferença agora evita horas de confusão depois.**

Em Python:

- Todo valor é um **objeto** que vive na memória.
- Uma variável é apenas um **nome (rótulo) apontando** para esse objeto.
- A atribuição `nome = valor` **não copia** nada: ela apenas gruda o rótulo no objeto.

```
    faturamento ──────► ┌──────────────┐
                        │  int  15000  │
    receita     ──────► └──────────────┘
```

Acima, `faturamento` e `receita` são dois nomes para o **mesmo** objeto.

### Regras de nomes

- Podem conter letras, números e `_`, mas **não podem começar com número**.
- São **case-sensitive**: `preco` e `Preco` são variáveis diferentes.
- Convenção Python (PEP 8): `snake_case` para variáveis e funções, `MAIUSCULAS` para constantes.
- Não use palavras reservadas (`if`, `for`, `class`, `None`, `lambda`...).

````python
# Atribuição simples
empresa = "Aurora Comércio"
cidade_sede = "Campinas-SP"
ano_fundacao = 2019
faturamento_mensal = 487_350.90   # o _ é só separador visual, Python ignora
ativa = True

print(empresa, "|", cidade_sede, "|", ano_fundacao)
print("Faturamento:", faturamento_mensal, "| Ativa:", ativa)
````

````python
# Dois nomes, um único objeto: prova com id()
faturamento = 15000
receita = faturamento

print("id(faturamento):", id(faturamento))
print("id(receita):    ", id(receita))
print("Mesmo objeto?", faturamento is receita)

# Rebind: 'faturamento' passa a apontar para OUTRO objeto.
# 'receita' continua apontando para o 15000 original.
faturamento = 20000
print()
print("faturamento =", faturamento)
print("receita     =", receita)
````

### Atribuição múltipla e troca de valores

Python permite atribuir vários nomes de uma vez. O idioma `a, b = b, a` é a forma canônica de trocar valores — sem variável temporária.

````python
# Vários nomes ao mesmo tempo
loja_a, loja_b, loja_c = 12000, 9500, 15300
print(loja_a, loja_b, loja_c)

# Mesmo valor para vários nomes
meta_q1 = meta_q2 = meta_q3 = 50000
print(meta_q1, meta_q2, meta_q3)

# Swap idiomático
maior, menor = 100, 5
maior, menor = menor, maior
print("Depois do swap:", maior, menor)
````

## 2. Tipos primitivos e `type()`

Python é de **tipagem dinâmica** (o tipo é do objeto, não do nome) e **forte** (ele não converte tipos incompatíveis silenciosamente — `"3" + 5` dá erro, e isso é bom).

| Tipo | Exemplo | Uso típico no Atlas |
|------|---------|---------------------|
| `int` | `42`, `-7`, `0` | Quantidade de itens, IDs de pedido |
| `float` | `3.14`, `19.90` | Preços, percentuais, médias |
| `str` | `"Aurora"`, `'SP'` | Nomes, cidades, categorias |
| `bool` | `True`, `False` | Flags: pedido pago? cliente ativo? |
| `NoneType` | `None` | Ausência de valor (campo não preenchido) |

`None` **não** é zero nem string vazia — é a ausência de valor. Guarde essa distinção: ela reaparece em bancos de dados como `NULL`.

````python
valores = [42, 19.90, "Aurora", True, None]

for v in valores:
    print(f"{repr(v):<10} -> {type(v).__name__}")
````

````python
# Conversão explícita (casting)
print(int("42") + 8)          # str -> int
print(float("19.90") * 2)     # str -> float
print(str(2026) + "-08-12")   # int -> str
print(int(19.99))             # float -> int  (TRUNCA, não arredonda!)
print(round(19.99))           # arredondamento correto
````

````python
# Tipagem FORTE: Python recusa a mistura ambígua.
# Descomente a linha abaixo e execute para ver o TypeError.

# print("3" + 5)

# A forma correta é ser explícito sobre a intenção:
print(int("3") + 5)    # 8   -> soma numérica
print("3" + str(5))    # 35  -> concatenação de texto
````

## 3. Números e operadores aritméticos

| Operador | Nome | Exemplo | Resultado |
|----------|------|---------|-----------|
| `+` | Soma | `7 + 2` | `9` |
| `-` | Subtração | `7 - 2` | `5` |
| `*` | Multiplicação | `7 * 2` | `14` |
| `/` | Divisão **real** | `7 / 2` | `3.5` (sempre `float`) |
| `//` | Divisão **inteira** | `7 // 2` | `3` |
| `%` | Resto (módulo) | `7 % 2` | `1` |
| `**` | Potência | `7 ** 2` | `49` |

A precedência segue a matemática: `**` primeiro, depois `* / // %`, depois `+ -`. **Na dúvida, use parênteses** — eles custam nada e economizam bugs.

````python
qtd = 7
preco = 129.90

print("Divisão real:    7 / 2  =", 7 / 2)
print("Divisão inteira: 7 // 2 =", 7 // 2)
print("Resto:           7 % 2  =", 7 % 2)
print("Potência:        7 ** 2 =", 7 ** 2)
print()
print("Total do pedido:", qtd * preco)
````

### O operador `%` (resto) na prática

O resto é muito mais útil do que parece. Três usos que você vai repetir a vida inteira:

- **Par ou ímpar:** `n % 2 == 0`
- **Múltiplo de N:** `n % N == 0`
- **"A cada N iterações":** dentro de um laço, `if i % 100 == 0: print(...)` para logar progresso.

````python
for n in [10, 15, 33, 100]:
    paridade = "par" if n % 2 == 0 else "ímpar"
    mult5 = "sim" if n % 5 == 0 else "não"
    print(f"{n:>4} -> {paridade:<5} | múltiplo de 5? {mult5}")
````

### Atribuição composta

`x += 1` é açúcar sintático para `x = x + 1`. Existe a versão composta de todos os operadores: `-=`, `*=`, `/=`, `//=`, `%=`, `**=`.

````python
carrinho = 0.0

carrinho += 129.90   # adiciona um mouse
carrinho += 349.00   # adiciona um teclado
carrinho *= 0.90     # aplica 10% de desconto
carrinho -= 15.00    # cupom de frete

print(f"Total do carrinho: R$ {carrinho:.2f}")
````

### ⚠️ Armadilha clássica: ponto flutuante

Execute a célula abaixo. O resultado vai te incomodar — e deve mesmo.

`float` é uma representação **binária** aproximada de números decimais. Isso não é bug do Python: é como praticamente toda linguagem representa números reais (padrão IEEE 754).

**Regra prática:** para dinheiro, nunca compare floats com `==`. Use `round()` na exibição e, quando a precisão for crítica (contabilidade, fiscal), use o módulo `decimal`.

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)          # False! 😱
print(round(0.1 + 0.2, 2) == 0.3)  # True

# Solução robusta para dinheiro:
from decimal import Decimal

print()
print(Decimal("0.1") + Decimal("0.2"))
print(Decimal("0.1") + Decimal("0.2") == Decimal("0.3"))
````

### Indexação e fatiamento (*slicing*)

```
 A   u   r   o   r   a
 0   1   2   3   4   5     <- índices positivos
-6  -5  -4  -3  -2  -1     <- índices negativos
```

A sintaxe do slice é `texto[inicio:fim:passo]`, onde **`fim` é exclusivo**.

````python
nome = "Aurora"

print("nome[0]    =", nome[0])
print("nome[-1]   =", nome[-1])
print("nome[0:3]  =", nome[0:3])   # 'Aur' — o índice 3 NÃO entra
print("nome[:3]   =", nome[:3])
print("nome[3:]   =", nome[3:])
print("nome[::2]  =", nome[::2])   # de 2 em 2
print("nome[::-1] =", nome[::-1])  # invertido
````

````python
# Strings são IMUTÁVEIS: você não altera, você cria uma nova.
nome = "aurora"

# nome[0] = "A"   # <- descomente para ver o TypeError

novo_nome = "A" + nome[1:]
print(novo_nome)
print(nome.capitalize())   # método já devolve uma nova string
````

### Métodos de string mais usados

Estes são os que você vai usar em 90% da limpeza de dados. **Todos devolvem uma nova string** — nenhum altera a original.

### f-strings: a forma moderna de formatar

Prefixe a string com `f` e coloque expressões entre `{}`. É mais legível e mais rápido que `%` ou `.format()`.

**Mini-linguagem de formatação** (vai dentro de `{valor:formato}`):

| Formato | Efeito | Exemplo de saída |
|---------|--------|------------------|
| `:.2f` | 2 casas decimais | `2599.90` |
| `:,.2f` | separador de milhar | `2,599.90` |
| `:>10` | alinha à direita em 10 col. | `␣␣␣␣Aurora` |
| `:<10` | alinha à esquerda | `Aurora␣␣␣␣` |
| `:^10` | centraliza | `␣␣Aurora␣␣` |
| `:.1%` | percentual | `12.5%` |
| `:=` (sufixo) | modo debug | `preco=2599.9` |

````python
produto = "Notebook Dell"
qtd = 3
preco = 2599.90
total = qtd * preco
margem = 0.235

print(f"{produto} x{qtd} = R$ {total:,.2f}")
print(f"Margem: {margem:.1%}")
print(f"{'Produto':<20}{'Qtd':>5}{'Total':>14}")
print(f"{produto:<20}{qtd:>5}{total:>14,.2f}")
print()
print(f"{preco=}")        # modo debug — ótimo para investigar
print(f"{qtd * preco=:.2f}")
````

## 6. Booleanos e operadores lógicos

### Comparação

| Operador | Significado |
|----------|-------------|
| `==` | igual a |
| `!=` | diferente de |
| `>` `<` | maior / menor |
| `>=` `<=` | maior ou igual / menor ou igual |
| `in` | pertence a |
| `is` | é **o mesmo objeto** |

⚠️ `==` compara **valor**; `is` compara **identidade**. Use `is` apenas com `None`, `True` e `False`.

````python
estoque = 3
minimo = 5

print("estoque == minimo :", estoque == minimo)
print("estoque != minimo :", estoque != minimo)
print("estoque <  minimo :", estoque < minimo)
print("'ELET' in 'AUR-ELET-01':", "ELET" in "AUR-ELET-01")

# Comparação encadeada — Python permite e é legível:
temperatura = 22
print("Faixa ideal (18-25)?", 18 <= temperatura <= 25)
````

````python
# == vs is
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print("a == b :", a == b)   # True  -> mesmo conteúdo
print("a is b :", a is b)   # False -> objetos diferentes na memória
print("a is c :", a is c)   # True  -> mesmo objeto

valor = None
print("Uso correto do is:", valor is None)
````

### `and`, `or`, `not` e a tabela-verdade

| A | B | `A and B` | `A or B` | `not A` |
|---|---|-----------|----------|---------|
| V | V | V | V | F |
| V | F | F | V | F |
| F | V | F | V | V |
| F | F | F | F | V |

**Curto-circuito:** em `A and B`, se `A` já é falso, `B` nem é avaliado. Em `A or B`, se `A` já é verdadeiro, `B` é ignorado. Isso permite escrever guardas seguras como `if pedido is not None and pedido.pago:`.

````python
pago = True
em_estoque = False
cliente_vip = True

print("pago and em_estoque      :", pago and em_estoque)
print("pago or  em_estoque      :", pago or em_estoque)
print("not em_estoque           :", not em_estoque)
print("pago and (em_estoque or cliente_vip):", pago and (em_estoque or cliente_vip))
````

### Truthy e Falsy

Qualquer objeto pode ser avaliado como condição. São considerados **falsos**:

`False`, `None`, `0`, `0.0`, `""` (string vazia), `[]`, `{}`, `()`, `set()`

**Todo o resto é verdadeiro** — inclusive `"False"`, `"0"` e `[0]`, que são strings/listas não vazias.

Por isso `if lista:` é mais idiomático que `if len(lista) > 0:`.

````python
casos = [0, 1, -1, 0.0, "", "0", "False", [], [0], {}, None, " "]

for caso in casos:
    print(f"{repr(caso):<10} -> {bool(caso)}")
````

## 7. Condicionais: `if` / `elif` / `else`

A estrutura de decisão. Em Python, **a indentação define o bloco** — não há chaves. O padrão da comunidade é **4 espaços**.

```python
if condicao:
    # executa se condicao for verdadeira
elif outra_condicao:
    # executa se a primeira for falsa E esta for verdadeira
else:
    # executa se nenhuma anterior for verdadeira
```

Só o **primeiro** ramo verdadeiro roda. `elif` e `else` são opcionais; você pode ter quantos `elif` quiser.

````python
valor_pedido = 320.00

if valor_pedido >= 500:
    frete = 0.0
    faixa = "Frete grátis"
elif valor_pedido >= 200:
    frete = 12.90
    faixa = "Frete promocional"
else:
    frete = 29.90
    faixa = "Frete padrão"

print(f"Pedido: R$ {valor_pedido:.2f}")
print(f"{faixa}: R$ {frete:.2f}")
print(f"Total:  R$ {valor_pedido + frete:.2f}")
````

````python
# Condições aninhadas: um if dentro de outro
cliente_vip = True
valor_pedido = 150.00

if cliente_vip:
    if valor_pedido >= 100:
        desconto = 0.15
    else:
        desconto = 0.05
else:
    if valor_pedido >= 300:
        desconto = 0.10
    else:
        desconto = 0.0

print(f"Desconto aplicado: {desconto:.0%}")
print(f"Valor final: R$ {valor_pedido * (1 - desconto):.2f}")
````

### Operador ternário

Para decisões simples de uma linha: `valor_se_verdadeiro if condicao else valor_se_falso`.

Use com moderação — se a expressão fica longa, um `if` normal é mais legível.

````python
estoque = 0

status = "Disponível" if estoque > 0 else "Esgotado"
print(status)

nota = 8.5
print("Aprovado" if nota >= 7 else "Reprovado")
````

### `match` / `case` (Python 3.10+)

Alternativa a longas cadeias de `elif` quando você compara **um valor contra várias possibilidades**. O `case _` é o "default".

````python
def descrever_status(status: str) -> str:
    match status:
        case "pago":
            return "✅ Pagamento confirmado — separar para envio"
        case "pendente":
            return "⏳ Aguardando pagamento"
        case "cancelado" | "estornado":
            return "❌ Pedido encerrado sem faturamento"
        case _:
            return "❓ Status desconhecido"


for s in ["pago", "pendente", "estornado", "zumbi"]:
    print(f"{s:<12} {descrever_status(s)}")
````

## 🔧 Prática guiada — Classificador de pedidos Aurora

Vamos juntar tudo. Este é o embrião do que você vai construir no `projeto_Atlas`.

````python
# Dados de um pedido (simulando uma linha de planilha)
pedido_id = 1042
cliente = "  maria SOUZA  "
cidade = "campinas"
produto = "Notebook Dell Inspiron 15"
quantidade = 2
preco_unitario = 2599.90
cupom = "AURORA10"
cliente_vip = True

# --- 1. Limpeza dos dados de texto ---
cliente_limpo = cliente.strip().title()
cidade_limpa = cidade.strip().title()

# --- 2. Cálculo do subtotal ---
subtotal = quantidade * preco_unitario

# --- 3. Regra de desconto ---
if cupom == "AURORA10":
    desconto_pct = 0.10
elif cliente_vip:
    desconto_pct = 0.05
else:
    desconto_pct = 0.0

valor_desconto = subtotal * desconto_pct
apos_desconto = subtotal - valor_desconto

# --- 4. Regra de frete ---
if apos_desconto >= 500:
    frete = 0.0
elif cidade_limpa == "Campinas":
    frete = 9.90
else:
    frete = 29.90

total = apos_desconto + frete

# --- 5. Faixa do pedido ---
if total >= 5000:
    faixa = "Alto valor"
elif total >= 1000:
    faixa = "Médio valor"
else:
    faixa = "Baixo valor"

# --- 6. Relatório ---
print("=" * 46)
print(f"{'PEDIDO ' + str(pedido_id):^46}")
print("=" * 46)
print(f"{'Cliente:':<16}{cliente_limpo}")
print(f"{'Cidade:':<16}{cidade_limpa}")
print(f"{'Produto:':<16}{produto}")
print("-" * 46)
print(f"{'Subtotal':<26}{subtotal:>20,.2f}")
print(f"{'Desconto (' + f'{desconto_pct:.0%}' + ')':<26}{-valor_desconto:>20,.2f}")
print(f"{'Frete':<26}{frete:>20,.2f}")
print("-" * 46)
print(f"{'TOTAL':<26}{total:>20,.2f}")
print(f"{'Classificação':<26}{faixa:>20}")
print("=" * 46)
````
